````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
print("База")
# 1. first task
print("1. first task")
cities_clean = [item.strip().lower().capitalize() for item in cities]
print("города: ", cities_clean)
# 2. second task
print("2. second task")
amounts_int = [int(item) for item in amounts]
print("список: ", amounts_int)
# 3. third task
print("3. third task")
counts = len(amounts_int) #общее число заказов
print("число заказов: ", counts) #общее число заказов
total = sum(amounts_int) #общая сумма
print("общая сумма: ", total) #общая сумма
avg = total / counts
print("средний чек: ", round(avg))
# 4. fourth task
print("4. fourth task")
unpaid = []
for u, n in zip(amounts_int, paid):
    if not n:
        unpaid.append(u)
    else:
        continue
print("неоплаченные суммы: ", unpaid) # неоплаченные суммы
print("кол-во неоплаченных сумм: ", len(unpaid))
print("общая неоплаченная сумма: ", sum(unpaid)) # неоплаченные суммы

# 5. fifth task
print("5. fifth task")
head = f"{'город:':>10} | {'сумма':>10} | статус"
print(head)
for location, summ, settled in zip(cities_clean, amounts_int, paid):
    status = "оплачено" if settled else "не оплачено"
    table = f"{location:>10} | {summ:>10} | {status}"
    print(table)
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт  
print("Стандарт")
# 6. sixth task
print("6. sixth task")
issued_astana = []
issued_almaty = []
for x, y, z in zip(amounts_int, paid, cities_clean):
    if y and z == "Astana":
        issued_astana.append(x)
    if y and z == "Almaty":
        issued_almaty.append(x)

print("Выручка по Астане (только оплаченные): ", sum(issued_astana))
print("Выручка по Алмате (только оплаченные): ", sum(issued_almaty))

# 7. seventh task
print("7. seventh task")
paid_amounts = []
paid_cities = []
for a, b, c in zip(amounts_int, paid, cities_clean):
    if b:
        paid_amounts.append(a)
        paid_cities.append(c)
    
max_amount = max(paid_amounts)
for f, e in zip(paid_amounts, paid_cities):
    if max_amount == f:
        max_city = e


max_index = amounts_int.index(max_amount)

print("выручка самого крупного оплаченного заказа: ", max_amount)
print("город самого крупного оплаченного заказа: ", max_city)
print("номер (индекс) самого крупного оплаченного заказа: ", max_index)

# 8. eighth task
print("8. eighth task")
weekends = [] 

for w in days:
    if w == "сб":
        weekends.append(w)
    if w == "вс":
        weekends.append(w)

print("Доля заказов выходного дня (`сб`, `вс`) в процентах: ", (len(weekends)/len(days)) * 100, "%")

# 9. ninth task
print("9. ninth task")
# Sn = (n * (2 * a1 + (n - 1)d)) / 2
accumulated = 0
increment = 15000
geom_inc = 1000
monthes = 0
g_accumulated = 0
g_monthes = 0

while accumulated < 400000:
    accumulated += increment
    monthes += 1
    increment = increment

while g_accumulated < 400000:
    g_accumulated += increment
    g_monthes += 1
    increment = increment + geom_inc

print("через сколько месяцев накопит в копилке если откладывать 15000 ежемесячно: ", monthes, "месяцев")
print("через сколько месяцев накопит в копилке если откладывать 15000 ежемесячно и на 1000 больше чем прошлый: ", g_monthes, "месяцев")
````
