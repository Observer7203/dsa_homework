````python
import csv
with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))
print(len(rows), "строк")
````

## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База

# 1. first task
print("1. first task")

CITY_MAP = { "Almaty": "Алматы", "Aktobe": "Актобе", "Shymkent": "Шымкент", "Astana": "Астана", "Karaganda": "Караганда", "Atyrau": "Атырау"}

def to_num(n):
    return float(n.replace(" ", ""))

def clean_city(n):
    n = n.strip()
    n = n.lower()
    n = n.capitalize()
    n = CITY_MAP.get(n, n)
    return n

cleaned_cities = {clean_city(r["city"]) for r in rows}

print("проверка функции clean_city(s): ", cleaned_cities)

print("проверка функции to_num(s): ", to_num("1 235"))

# 2. second task
print("2. second task")

cat_counts = {}
for r in rows:
    cat = r["category"]
    cat_counts[cat] = cat_counts.get(cat, 0) + 1

print(f"{'категория':<12} | заказов")
for cat, n in sorted(cat_counts.items(), key=lambda t: t[1], reverse=True):
    print(f"{cat:<12} | {n}")

# 3. third task
print("3. third task")

cat_revenue = {}
cat_paid = {}

for r in rows:
    if r["status"] == "оплачен":
        cat = r["category"]
        cat_revenue[cat] = cat_revenue.get(cat, 0) + to_num(r["amount"])
        cat_paid[cat] = cat_paid.get(cat, 0) + 1

print(f"{'категория':<12} | {'выручка':>12} | {'ср. чек':>8}")

for cat in cat_revenue:
    avg = cat_revenue[cat] / cat_paid[cat]
    print(f"{cat:<12} | {cat_revenue[cat]:>12,.0f} | {avg:>8,.0f}")

# 4. fourth task
print("4. fourth task")

month_counts = {}
for r in rows:
    month = r["order_date"][:7]
    month_counts[month] = month_counts.get(month, 0) + 1

print(f"{'месяц':<8} | заказов")
for month, n in sorted(month_counts.items()):
    print(f"{month:<8} | {n}")


````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
# Стандарт

# 5. fifth task
print("5. fifth task")

month_revenue = {}
for r in rows:
    if r["status"] == "оплачен":
        month = r["order_date"][:7]
        month_revenue[month] = month_revenue.get(month, 0) + to_num(r["amount"])

print(f"{'месяц':<8} | {'выручка':>12}")
for month, rev in sorted(month_revenue.items()):
    print(f"{month:<8} | {rev:>12,.0f}")

sep_rev = month_revenue["2026-09"]
oct_rev = month_revenue["2026-10"]
change = (oct_rev - sep_rev) / sep_rev * 100
print(f"Изменение октября к сентябрю: {change:.1f}%")

# 6. sixth task
print("6. sixth task")

city_total = {}      
city_cancelled = {}  
for r in rows:
    city = clean_city(r["city"])
    city_total[city] = city_total.get(city, 0) + 1
    if r["status"] == "отменён":
        city_cancelled[city] = city_cancelled.get(city, 0) + 1

print(f"{'город':<10} | {'заказов':>7} | {'отменено':>8} | доля")
for city in city_total:
    cancelled = city_cancelled.get(city, 0)
    share = cancelled / city_total[city] * 100
    print(f"{city:<10} | {city_total[city]:>7} | {cancelled:>8} | {share:.1f}%")

# 7. seventh task
print("7. seventh task")

cust_revenue = {}    
for r in rows:
    if r["status"] == "оплачен":
        cid = r["customer_id"]
        cust_revenue[cid] = cust_revenue.get(cid, 0) + to_num(r["amount"])

top5 = sorted(cust_revenue.items(), key=lambda t: t[1], reverse=True)[:5]

print(f"{'клиент':<8} | {'выручка':>12}")
for cid, rev in top5:
    print(f"{cid:<8} | {rev:>12,.0f}")

# 8. eighth task
print("8. eighth task")

city_customers = {}    
for r in rows:
    city = clean_city(r["city"])
    city_customers.setdefault(city, set()).add(r["customer_id"])

print(f"{'город':<10} | уникальных клиентов")
for city, customers in sorted(city_customers.items(), key=lambda t: len(t[1]), reverse=True):
    print(f"{city:<10} | {len(customers)}")



````

## [Челлендж]
9. Найдите полные дубли строк: превратите каждую строку в `tuple(r.values())` и проверяйте через множество. Сколько их и в каком месяце?
10. Пересчитайте выручку по месяцам без дублей и без клиента из задачи 7. Как изменился процент из задачи 5?
11. У какого клиента больше всего заказов (после удаления дублей)? Сколько?
12. Напишите менеджеру 3–4 предложения: насколько на самом деле упала выручка и почему первый расчёт был неверным.

````python
# Челлендж

# 9. ninth task
print("9. ninth task")

seen = set() 
dups = []        
for r in rows:
    key = tuple(r.values())
    if key in seen:
        dups.append(r)
    else:
        seen.add(key)

print("Полных дублей:", len(dups))

dup_months = {}
for r in dups:
    month = r["order_date"][:7]
    dup_months[month] = dup_months.get(month, 0) + 1
print("Дубли по месяцам:", dup_months)

# 10. tenth task
print("10. tenth task")


seen = set()
clean_rows = []
for r in rows:
    key = tuple(r.values())
    if key in seen:                 
        continue
    seen.add(key)
    if r["customer_id"] == "TEST": 
        continue
    clean_rows.append(r)

print("Строк после очистки:", len(clean_rows))

clean_revenue = {}
for r in clean_rows:
    if r["status"] == "оплачен":
        month = r["order_date"][:7]
        clean_revenue[month] = clean_revenue.get(month, 0) + to_num(r["amount"])

print(f"{'месяц':<8} | {'выручка':>12}")
for month, rev in sorted(clean_revenue.items()):
    print(f"{month:<8} | {rev:>12,.0f}")

sep_rev = clean_revenue["2026-09"]
oct_rev = clean_revenue["2026-10"]
new_change = (oct_rev - sep_rev) / sep_rev * 100
print(f"Изменение октября к сентябрю: было {change:.1f}%, стало {new_change:.1f}%")

# 11. eleventh task
print("11. eleventh task")

cust_orders = {}       
for r in clean_rows:
    cid = r["customer_id"]
    cust_orders[cid] = cust_orders.get(cid, 0) + 1

best = max(cust_orders, key=cust_orders.get)
print("Больше всего заказов у клиента:", best, "—", cust_orders[best])


````
